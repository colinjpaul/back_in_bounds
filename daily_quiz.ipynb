understanding timing decorator

```python
import time
import functools

from functools import wraps

def timed(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(f"{func.__name__} took {time.perf_counter() - start:.3f}s")
        return result
    return wrapper

@timed
def total_score(scores):
    """Sum a list of hole scores."""
    return sum(scores)

print(total_score([4, 5, 3]))
print(total_score.__name__)
print(total_score.__doc__)
```
