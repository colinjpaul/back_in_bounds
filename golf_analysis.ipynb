```python
import pandas as pd

def clean_and_convert_dates(date_series):
    # 1. Standardise separators: replace all slashes with dashes
    clean_series = date_series.astype(str).str.replace('/', '-', regex=False)
    
    # 2. Convert to datetime with a safety net (errors='coerce')
    return pd.to_datetime(clean_series, format='%d-%m-%y', errors='coerce')

# --- TEST IT OUT ---
# A messy mix of dashes, slashes, and a typo ('N/A')
mixed_data = pd.Series(['06-07-24', '21/05/26', '23-05-26', 'garbage_date'])

converted_dates = clean_and_convert_dates(mixed_data)
print(converted_dates)
```
